```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.pyplot as plt
import logging
logging.getLogger('matplotlib.font_manager').disabled = True

df= pd.read_csv("/content/Updated_NSQIP_83083.csv")

# 1. Setup Plot Style
sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'Segoe UI'

# 2. Create the Figure
fig, ax = plt.subplots(figsize=(12, 6.5)) # Roughly 902x507 pixels

# Define the palette with lowercase keys to match the error message
custom_palette = {'female': '#1f77b4', 'male': '#ff7f0e'}

# 3. Create the Boxplot
# Note: JMP "Overlay" maps to 'hue' in Seaborn
sns.boxplot(
    data=df,
    x='PUFYEAR',
    y='OPTIME',
    hue='SEX',
    palette=custom_palette, # Use the corrected palette
    width=0.6,
    ax=ax
)

# 4. Add Mean Points (The JMP "Points" element with Mean Summary)
sns.pointplot(
    data=df,
    x='PUFYEAR',
    y='OPTIME',
    hue='SEX',
    dodge=0.5,           # Aligns points with the boxplot centers
    linestyle='none',    # No lines connecting means (replaces join=False)
    markers='D',         # Diamond shape for means
    markersize=8,        # Size of markers (replaces scale=0.75)
    errorbar=None,       # Remove error bars for cleaner look
    palette=custom_palette, # Apply the same palette to pointplot
    ax=ax
)

# 5. Formatting (The "SendToReport" logic)
ax.set_yscale('log')     # Log Scale for OPTIME
ax.set_title("Operation Time by Year and Sex", fontsize=14, fontweight='semibold')
ax.set_xlabel("", fontsize=12) # Hidden per JMP code
ax.set_ylabel("Operation Time (min)", fontsize=12)

# Customize Legend
handles, labels = ax.get_legend_handles_labels()
# Only take the first two handles (from boxplot) to avoid duplicates from pointplot
# Update labels to match the lowercase keys for consistency in legend
updated_labels = [label.capitalize() for label in labels[0:2]]
ax.legend(handles[0:2], updated_labels, title='Sex', loc='upper right', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import logging

logging.getLogger('matplotlib.font_manager').disabled = True

# Load Data
df = pd.read_csv("/content/Updated_NSQIP_83083.csv")

# 1. Setup Plot Style
sns.set_theme(style="white") # 'white' better matches JMP background
plt.rcParams['font.family'] = 'Segoe UI'

# 2. Create the Figure
fig, ax = plt.subplots(figsize=(12, 6.5))

# Use the colors from your JMP image: Blue for Female, Red/Pink for Male
custom_palette = {'female': '#4C72B0', 'male': '#C44E52'}

# 3. Create the Boxplot
sns.boxplot(
    data=df, x='PUFYEAR', y='OPTIME', hue='SEX',
    palette=custom_palette, width=0.5, ax=ax,
    fliersize=0, # Hide default outliers to avoid clutter
    boxprops=dict(facecolor='none', linewidth=1) # Hollow boxes
)

# 4. Add Dense Outliers (The "Mountain" effect)
sns.stripplot(
    data=df, x='PUFYEAR', y='OPTIME', hue='SEX',
    palette=custom_palette, dodge=True, jitter=0.3,
    size=2, alpha=0.3, zorder=1, ax=ax, legend=False
)

# 5. Add Value Labels (JMP "Label by Value" style)
# Calculate mean for each group
means = df.groupby(['PUFYEAR', 'SEX'])['OPTIME'].mean().reset_index()
years = sorted(df['PUFYEAR'].unique())

for i, year in enumerate(years):
    # Filter for each sex to place labels
    for sex_val in ['female', 'male']:
        mean_val = means[(means['PUFYEAR'] == year) & (means['SEX'] == sex_val)]['OPTIME'].values[0]

        # Position offset: -0.15 for female box, +0.15 for male box
        x_offset = -0.15 if sex_val == 'female' else 0.15

        ax.text(
            i + x_offset, mean_val * 1.1, # Placed slightly above the mean point
            f'{mean_val:.1f}',
            color=custom_palette[sex_val],
            ha='center', va='bottom',
            fontsize=10, fontweight='bold',
            bbox=dict(facecolor='white', alpha=0.7, edgecolor='none', pad=0.5)
        )

# 6. Formatting
ax.set_yscale('log')
ax.set_title("Operation Time by Year and Sex", fontsize=14, fontweight='semibold')
ax.set_xlabel("", fontsize=12)
ax.set_ylabel("Operation Time (min)", fontsize=12)

# Set specific Y-ticks to match JMP's log scale (10, 20, 30... 1000)
ax.set_yticks([1, 2, 5, 10, 20, 30, 40, 50, 100, 200, 300, 400, 500, 1000])
ax.get_yaxis().set_major_formatter(plt.ScalarFormatter())

# Clean up Legend
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles[0:2], [l.capitalize() for l in labels[0:2]], title='SEX', loc='upper right')

sns.despine() # Matches JMP's clean top/right look
plt.tight_layout()
plt.show()
```
